# NutriSense reproducible analysis

This notebook is a thin, clean-kernel wrapper around the canonical CLI. Historical report values are not inputs. Synthetic runs must use explicit CLI paths and are never copied into reports.

```python
from pathlib import Path
import subprocess
import sys

cwd = Path.cwd().resolve()
repo = next(path for path in (cwd, *cwd.parents) if (path / 'analysis' / 'run_analysis.py').exists())
command = [sys.executable, str(repo / 'analysis' / 'run_analysis.py'), '--mode', 'real']
completed = subprocess.run(command, cwd=repo, text=True, capture_output=True, check=False)
print(completed.stdout)
assert completed.returncode == 0, completed.stderr
```

```python
import json

manifest = json.loads((repo / 'analysis' / 'results_manifest.json').read_text(encoding='utf-8'))
print('STATUS:', manifest['status'])
print('synthetic:', manifest['synthetic'])
print('result_count:', len(manifest.get('results', {})))
```
